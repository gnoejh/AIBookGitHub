# Generative AI: From Pixels to Consistency

Welcome to the advanced Generative AI module. This series of notebooks covers the rapid evolution of image generation technologies, moving from foundational probabilistic models to state-of-the-art real-time generation.

## 📚 Table of Contents

1.  **[The Foundation: Denoising Diffusion Probabilistic Models (DDPM)](./1_ddpm.ipynb)**
    *   *Concept:* Learning to reverse the noise process pixel-by-pixel.
    *   *Key Takeaway:* High-quality generation is possible via iterative denoising.

2.  **[The Eye: CLIP (Contrastive Language-Image Pre-training)](./2_clip.ipynb)**
    *   *Concept:* Connecting text and images in a shared vector space.
    *   *Key Takeaway:* How models understand "A futuristic city" to guide generation.

3.  **[The Efficiency: Latent Diffusion Models (LDM)](./3_ldm.ipynb)**
    *   *Concept:* Moving diffusion from pixel space to compressed latent space.
    *   *Key Takeaway:* High-resolution generation on consumer hardware (Stable Diffusion).

4.  **[The Backbone: Diffusion Transformers (DiT)](./4_dit.ipynb)**
    *   *Concept:* Replacing the U-Net with a Scalable Transformer.
    *   *Key Takeaway:* Scaling laws apply to image generation (Sora, SD3).

5.  **[The Path: Flow Matching](./5_flow_matching.ipynb)**
    *   *Concept:* Straightening the generation path from noise to data.
    *   *Key Takeaway:* Faster, more stable training and inference (Flux, SD3).

6.  **[The Speed: Consistency Models](./6_consistency_models.ipynb)**
    *   *Concept:* Mapping any point on the trajectory directly to the solution.
    *   *Key Takeaway:* Real-time, 1-step to 4-step generation.


## 🚀 The Evolution of Generative AI

The field has moved towards **Efficiency**, **Scalability**, and **Speed**.

1.  **DDPM (2020):** Proved diffusion works better than GANs for mode coverage, but was slow and expensive.
2.  **CLIP (2021):** Solved the "control" problem, allowing text-to-image.
3.  **LDM (2022):** Solved the "efficiency" problem, democratizing access.
4.  **DiT (2023):** Solved the "scaling" problem, enabling massive models.
5.  **Flow Matching (2023):** Solved the "path" problem, optimizing the math.
6.  **Consistency (2023/24):** Solved the "speed" problem, enabling real-time apps.

## 📊 Comparison of Architectures

| **Training** | Noise Prediction | Noise Prediction | Noise/Velocity | **Vector Field / Velocity** | Consistency Mapping |

| **Generation Speed** | Very Slow | Moderate | Moderate | Fast | **Real-Time** |
| **Scalability** | Limited | Good | **Excellent** | Excellent | Good |
